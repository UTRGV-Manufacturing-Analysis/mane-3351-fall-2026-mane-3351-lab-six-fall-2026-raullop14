```python
# Cell 2
import math
import numpy as np
import matplotlib.pyplot as plt

# Step 2 - insert the user-defined function from Lab Five that calculates the CDF of the logistic distribution below this line

def logistic_cdf(x, mu, sigma):
    return 1 / (1 + math.exp(-(x - mu) / sigma))

# Step 3 - Insert the Python command(s) to vectorize your user-defined function from Step 1

vectorized_logistic_cdf = np.vectorize(logistic_cdf)

# Step 4 - create array containing 500 points (using the linspace Numpy command) for values of x from -10 to +10

x= np.linspace(-10, 10, 500)

# Step 5 - create the array y using the vectorized CDF for the logistic distribution with mu=2.0 and sigma=5.0 created in Step 2

y= vectorized_logistic_cdf(x, 2.0, 5.0)

# Step 6 - using the code below to create a graph of the data created in Steps 3 and 4

plt.plot(x,y,'b')
plt.xlabel('x')
plt.ylabel('CDF')
plt.title('Logistic Distribution CDF (mu=2.0, sigma=5.0)')
plt.grid(True)
plt.show()
```
